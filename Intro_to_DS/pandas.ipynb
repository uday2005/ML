```python

import pandas as pd
import numpy as np
def read_series():
    values = []
    indexs = []
    while True:
        ll = input()
        if not  ll:
            break
        ll = ll.split()
        try:
            values.append(ll[1])
            indexs.append(ll[0])
        except:
            print("out of index")
        
    

    return pd.Series(values , index=indexs )
```
