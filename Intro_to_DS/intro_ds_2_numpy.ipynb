concatenate and stack. Concatenate takes n-dimensional arrays and <br>
returns an n-dimensional array, whereas stack takes <br>
 n-dimensional arrays and returns n+1-dimensional array.

```python
a=np.arange(2)
b=np.arange(2,5)
print(f"a has shape {a.shape}: {a}")
print(f"b has shape {b.shape}: {b}")
np.concatenate((a,b))  # concatenating 1d arrays

c=np.arange(1,5).reshape(2,2)
print(c)
print(f"c has shape {c.shape}:", c, sep="\n")
np.concatenate((c,c))   # concatenating 2d arrays

# By default concatenate joins the arrays along axis 0. To join the arrays horizontally, add parameter axis=1

np.concatenate((c,c), axis=1)

# axis=0 → "down" (add rows)
# axis=1 → "across" (add columns)

# Use stack to create higher dimensional arrays from lower dimensional arrays

np.stack((b,b))

np.stack((b,b), axis=1)
```

```python
import numpy as np

def vector_angles(X, Y):
    # Dot product for each pair of vectors
    dot_product = np.sum(X * Y, axis=1)
    # Norms for each vector
    norm_X = np.linalg.norm(X, axis=1)
    norm_Y = np.linalg.norm(Y, axis=1)
    # Cosine of angles
    cos_angles = dot_product / (norm_X * norm_Y)
    # Clip for numerical stability
    cos_angles = np.clip(cos_angles, -1.0, 1.0)
    # Angle in radians
    angles_rad = np.arccos(cos_angles)
    # Convert to degrees
    angles_deg = np.degrees(angles_rad)
    return angles_deg

A=np.array([[0,0,1], [-1,1,0]])
B=np.array([[0,1,0], [1,1,0]])
a = vector_angles(A, B)

print(a)
```

Broadcasting <br>
We have seen that NumPy allows array operations that are performed element-wise.  <br>
But NumPy also allows binary operations that don't require the two arrays to have the same shape.

```python
np.arange(3) + np.array([4])
```

NumPy lines up shapes by adding 1’s to the left.

Each dimension must be either the same or 1.

If it’s 1, that value is repeated as needed.

If not, you get an error.
