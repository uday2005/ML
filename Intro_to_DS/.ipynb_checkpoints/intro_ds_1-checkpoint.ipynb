```python
# in the common UTF-8 encoding the character c corresponds to the byte with integer value 99 and
# the character ä corresponds to sequence of bytes [195, 164].
# An example conversion between characters and bytes
```

```python
b="ä".encode("utf-8")     # Convert character(s) to a sequence of bytes
print(b)                  # Prints bytes in hexadecimal notation
print(list(b))            # Prints bytes in decimal notation
```

```python
bytes.decode(b, "utf-8")  # convert sequence of bytes to character(s)
```
