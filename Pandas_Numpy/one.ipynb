When you find a book with missing pages (NaN), you have three main strategies: <br>
Deal with it (Ignore): If you're just making a list of book titles, maybe you don't care if a few pages are missing  <br> inside. For some analyses, this is okay. But for modeling, it's rarely an option. <br>
Drop it (Discard): If a book is mostly just a cover with almost all pages missing (like our Cabin column), it's probably  <br> useless. It's best to drop the entire book (the column). If a book only has one or two non-critical pages missing, <br> you might decide to discard that single book (the row). <br>
Fill it (Impute): This is often the best and most common strategy. You can't magically know what was on the missing page, but you can make an educated guess. This is called imputation.<br>
For a missing chapter number (a numerical value), a simple guess is to fill it in with the average or median chapter number from all the other books. <br>
For a missing book category (a categorical value), a simple guess is to fill it in with the most common category (the mode).<br>

```python
print("\n--- Step 2: Dropping a column with too many missing values ---")

# The Cabin column is over 75% missing (687 / 891). Trying to guess these values would be pure fiction. 
# The data it contains is likely not reliable enough to use. The best strategy is to remove it completely.
print(df_clean.columns)

# axis=1 tells pandas to drop a column. axis=0 would mean drop a row.
df_clean = df_clean.drop('Cabin', axis=1)

# print("Cabin column dropped. Let's check the columns again:")
```

```python
# Step 3: Filling 'Embarked' with the mode
print("\n--- Step 3: Filling missing categorical data with the mode ---")

# First, find the mode. .mode() returns a Series because there could be a tie,
# so we take the first element [0].
embarked_mode = df_clean['Embarked'].mode()[0]
print(f"The most common embarkation port (the mode) is: {embarked_mode}")

# Now, fill the NaNs with this mode.
# We'll use inplace=True here for convenience.
df_clean['Embarked'].fillna({'Embarked':embarked_mode}, inplace=True)

# Verify the fix
print(f"\nMissing values in 'Embarked' after filling: {df_clean['Embarked'].isnull().sum()}") # Should be 0!
```

```python
# Step 4: Filling 'Age' with the median
print("\n--- Step 4: Filling missing numerical data with the median ---")

age_median = df_clean['Age'].median()
print(f"The median age is: {age_median}")

df_clean['Age'] = df_clean['Age'].fillna(age_median)

# Verify the fix
print(f"Missing values in 'Age' after filling: {df_clean['Age'].isnull().sum()}") # Should also be 0!

print(df_clean.isnull().sum())
```

```python
import pandas as pd
df = pd.read_csv('train.csv')

# --- Quick Cleaning from Week 4 ---

df_clean = df.copy()
df_clean = df_clean.drop('Cabin', axis=1) # Drop Cabin
embarked_mode = df_clean['Embarked'].mode()[0]
df_clean['Embarked'] = df_clean['Embarked'].fillna(embarked_mode) # Fill Embarked
age_median = df_clean['Age'].median()
df_clean['Age'] = df_clean['Age'].fillna(age_median) # Fill Age
# --- End of Cleaning ---

print("Clean DataFrame is ready. Let's begin alchemy!")
```

```python
# Step 1: Creating 'FamilySize'
print("\n--- Step 1: Creating the 'FamilySize' column ---")

df_clean['FamilySize'] = df_clean['SibSp'] + df_clean['Parch'] + 1

# Let's check our work. We can use .loc to see the original columns and our new one side-by-side.
print("Verifying the new 'FamilySize' column:")
print(df_clean.loc[:, ['SibSp', 'Parch', 'FamilySize']].head())
```

```python
# Step 2: Creating the 'IsAlone' column
print("\n--- Step 2: Creating a boolean 'IsAlone' column ---")

# The expression (df_clean['FamilySize'] == 1) will result in a Series of True or False values.
df_clean['IsAlone'] = (df_clean['FamilySize'] == 1)

print("Verifying the new 'IsAlone' column:")
print(df_clean.loc[:, ['FamilySize', 'IsAlone']].head())
```

```python
# Step 3: Sorting by Fare to find the richest passengers
print("\n--- Step 3: Sorting by a single column ('Fare') ---")

richest_passengers = df_clean.sort_values(by='Fare', ascending=False)

# Now let's look at the top 5 from this sorted DataFrame
print("The 5 passengers who paid the highest fares:")
print(richest_passengers[['Name', 'Pclass', 'Fare']].head())
```

```python
# Step 4: Sorting by multiple columns
print("\n--- Step 4: Sorting by multiple columns ('Pclass' then 'Age') ---")

# We provide a list of column names and a corresponding list for ascending/descending
sorted_by_class_and_age = df_clean.sort_values(by=['Pclass', 'Age'], ascending=[True, False])

print("Passengers sorted by class, then by age (descending):")
print(sorted_by_class_and_age[['Name', 'Pclass', 'Age']].head(10))
```

```python
# Let's start with our clean DataFrame from Week 5
import pandas as pd
import numpy as np
# (Assuming df_clean is ready)

# Step 1: Survival Rate by Sex
print("\n--- Step 1: Calculating Survival Rate by Sex ---")

# 1. SPLIT the data into a 'male' basket and a 'female' basket.
# 2. APPLY the .mean() function to the 'Survived' column within each basket.
#    (Because Survived is 0 or 1, the mean is the same as the survival percentage!)
# 3. COMBINE the results into a new Series.
survival_by_sex = df_clean.groupby('Sex')['Survived'].mean()

print(survival_by_sex)
```

```python
# Step 2: Analyzing Age and Fare by Passenger Class
print("\n--- Step 2: Median Age and Average Fare by Class ---")

# Group by 'Pclass', then select the 'Age' and 'Fare' columns.
# We can then ask for an aggregation on this selection. Let's get the median.
class_analysis = df_clean.groupby('Pclass')[['Age', 'Fare']].median()

print(class_analysis)
```

```python
# Step 3: Multi-level Grouping (Class AND Sex)
print("\n--- Step 3: Survival Rate by Class AND Sex ---")

# To group by multiple columns, we pass a list of column names.
survival_by_class_sex = df_clean.groupby(['Pclass', 'Sex'])['Survived'].mean()

print(survival_by_class_sex)
```

```python
# Step 4: Advanced Aggregation with .agg()
print("\n--- Step 4: Creating a custom summary with .agg() ---")

# We pass a dictionary where keys are the new column names in our summary,
# and values are a tuple: ('original_column_name', 'function_to_apply')
adv_analysis = df_clean.groupby('Pclass').agg(
    Total_Passengers=('PassengerId', 'count'), # Count passengers in each class
    Total_Survivors=('Survived', 'sum'),      # Sum of 1s is the count of survivors
    Median_Age=('Age', 'median'),             # Get the median age
    Average_Fare=('Fare', 'mean')             # Get the average fare
)

print(adv_analysis)
```
