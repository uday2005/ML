```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# This dataset is a classic and is built into every Seaborn version.
# This command is guaranteed to work.
df = sns.load_dataset('diamonds')

# Let's inspect our new dataset
print("--- Successfully loaded the 'diamonds' dataset ---")
print("\n--- First 5 Rows ---")
print(df.head())

print("\n--- Data Info ---")
df.info()

print(df.describe()) # Some value of x y and z are negative which is not possible in reality so we removing it.

df = df[(df['x']>0) & (df['y'] >0) & (df['z'] >0)] # removing those value which are less than zero.

```

```python
fig , axes = plt.subplots(1,2 ,figsize=(15,12))
ax1 = axes[0]
ax2 = axes[1]

# First i want to see the ditribution so that i will get to know more about data liek is it skewed or not
sns.histplot(data= df,ax=ax1,x='price',color='red',kde=True)
sns.histplot(data= df,ax=ax2,x='carat',color='blue',kde=True)
plt.tight_layout()
plt.show()
```

"The distribution of diamond prices is heavily right-skewed, indicating that while most diamonds are affordable, a small number of them have exceptionally high prices. Carat weight is similarly skewed, suggesting most diamonds in the market are small.

```python
fig,ax = plt.subplots(figsize=(15,12))

sns.scatterplot(data = df, x='carat' ,y='price' ,hue='clarity',palette='viridis', ax= ax,alpha=0.7)
ax.set_title('realtionship b/w carat and price with color coded clarity')

```

"There is a strong, positive, and non-linear relationship between carat and price. As carat increases, price increases exponentially. Within any given carat weight, diamonds with better clarity (higher on the color scale) tend to command a higher pric

```python
fig,ax = plt.subplots(figsize=(15,12))
cut_order = ['Fair', 'Good', 'Very Good', 'Premium', 'Ideal']
sns.violinplot(data= df,x='cut' , y='price',order=cut_order,palette='viridis',hue='cut',legend=False)

```

While one might expect 'Ideal' cut diamonds to have the highest median price, the distributions are heavily overlapping. In fact, 'Premium' and 'Very Good' cuts often have a higher median price, suggesting that cut quality alone is not the sole determinant of price and its effect is complex."
