You're absolutely correct! In Matplotlib, you cannot directly pass a DataFrame as an axis to plotting functions. Instead, you need to extract the specific columns or data series from the DataFrame that you want to plot.

so we use seaborn for that as it can take the dataframe directly


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
tips = sns.load_dataset('tips')
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

sns.scatterplot(data=tips, x='total_bill' , y='tip' , ax = ax)

ax.set_title("Total_bill vs Tip comparsion")
ax.set_xlabel("Total_bill")

plt.show()
```

Axes-level functions: scatterplot(), histplot(), barplot(), boxplot(), etc. <br>
These functions draw a plot on a single Matplotlib Axes. <br>
They are identifiable because they take an ax= argument. <br>
This is what we will use. It integrates perfectly with our fig, ax = plt.subplots() workflow. <br>
Figure-level functions: relplot() (for relationships), displot() (for distributions), catplot() (for categories). <br>
These functions create their own Figure and Axes, often creating a grid of subplots (faceting). <br>
They are more powerful for complex comparisons but don't take an ax= argument

```python
sns.set_theme(style="ticks", context="talk")

# Load the dataset
tips = sns.load_dataset('tips')
```

```python
# Create the figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Let Seaborn do the hard work of plotting the bar chart.
# It will automatically calculate the mean of 'total_bill' for each 'day'.
sns.barplot(data=tips, x='day', y='total_bill', ax=ax)

# Add a title using our Matplotlib skills
ax.set_title('Average Total Bill by Day of the Week')
ax.set_xlabel('Day of the Week')
ax.set_ylabel('Average Total Bill ($)')

plt.show()
```

```python
# Create the figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Use Seaborn's histplot
# The `kde=True` adds a smooth line representing the estimated distribution
sns.histplot(data=tips, x='total_bill', kde=True, ax=ax)

# Add a title
ax.set_title('Distribution of Total Bill Amounts')
ax.set_xlabel('Total Bill ($)')
ax.set_ylabel('Count (Number of Tables)')

plt.show()
```

```python
# Create the figure and axes
fig, ax = plt.subplots(figsize=(10, 6))

# Use Seaborn's scatterplot
sns.scatterplot(data=tips, x='total_bill', y='tip', ax=ax)

# Add a title
ax.set_title('Relationship between Total Bill and Tip')
ax.set_xlabel('Total Bill ($)')
ax.set_ylabel('Tip Amount ($)')


plt.show()
```

```python
fig, ax = plt.subplots(figsize=(12, 7))

# A boxplot is much clearer for this kind of relationship
sns.boxplot(data=tips, x='size', y='tip', ax=ax)

# REVISED, ACCURATE LABELS
ax.set_title('Relationship Between Party Size and Tip Amount', fontsize=16, fontweight='bold')
ax.set_xlabel('Size of Party', fontsize=12)
ax.set_ylabel('Tip Amount ($)', fontsize=12)
ax.spines[['top', 'right']].set_visible(False)
plt.show()
```

```python
# "How do the total bills compare between smokers and non-smokers?" -> Bar Plot.
# Correct. This is a Comparison problem. You are comparing a numerical aggregate (the average total bill) across two distinct categories ('smoker' vs. 'non-smoker'). 
# A bar plot is the ideal tool for this. A box plot would also be an excellent choice as it shows more of the distribution, but a bar plot is a direct and clear answer to the question.
    
# "What is the distribution of tip amounts?" -> Histogram.
# Correct. This is a classic Distribution problem. You want to see the shape, frequency, and spread of a single continuous variable ('tip'). A histogram is the definitive tool for this job.

# "Is there a relationship between the size of the party (size) and the tip amount?" -> Scatter Plot.
# Correct. This is a Relationship problem. You are examining how one numerical variable ('tip') changes with another numerical variable ('size'). A scatter plot is the perfect way to visualize this relationship and see if a trend exists.
```
