```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

titanic = sns.load_dataset('titanic')

# --- PLOT 1: Survival Rate by Class ---
fig, ax = plt.subplots(figsize=(10, 6))

# Let barplot calculate the mean of 'survived' (the survival rate)
# We put the categorical variable on the y-axis to make it horizontal.
sns.barplot(data=titanic, y='pclass', x='survived', orient='h', ax=ax)

# Polish the plot
ax.set_title('Survival Rate Was Highest for 1st Class Passengers', fontsize=16, fontweight='bold')
ax.set_xlabel('Survival Rate', fontsize=12)
ax.set_ylabel('Passenger Class', fontsize=12)
ax.set_xlim(0, 1) # Survival rate is a percentage, so it's between 0 and 1.
ax.spines[['top', 'right']].set_visible(False)

# Add data labels to the bars for clarity
for p in ax.patches:
    width = p.get_width()
    ax.text(width + 0.01, p.get_y() + p.get_height() / 2,
            f'{width:.1%}', # Format as percentage
            ha='left', va='center')

plt.show()
```

```python
# Create a pivot table to get the composition
composition = titanic.pivot_table(index='pclass', columns='sex', aggfunc='size')

print(composition)
fig, ax = plt.subplots(figsize=(10, 7))

composition.plot(kind='bar', stacked=True, ax=ax, colormap='viridis')

# Polish the plot
ax.set_title('Gender Composition by Passenger Class', fontsize=16, fontweight='bold')
ax.set_xlabel('Passenger Class', fontsize=12)
ax.set_ylabel('Number of Passengers', fontsize=12)
ax.spines[['top', 'right']].set_visible(False)
ax.tick_params(axis='x', rotation=0) # Keep the x-labels horizontal

plt.show()
```

```python
tips = sns.load_dataset('tips')

fig , ax = plt.subplots(figsize=(10,7))

parameter = tips.groupby('day')['total_bill'].mean().sort_values(ascending=False).index
# as all these would have been data frame with index as day adn we order the barplot by day so we need index of pandas dataframe

sns.barplot(data=tips, x='day' , y='total_bill',ax=ax,order=parameter)

ax.spines[['top', 'right']].set_visible(False)
ax.set_title('Total bill spended on a particular day' , fontweight='bold', fontsize=16)
ax.set_xlabel('Day',fontweight=12)
ax.set_ylabel('Total bill' , fontweight=12)

# Assume 'ax' is your axes object with the bar plot already drawn on it.

# Loop through each bar (patch) in the plot
for p in ax.patches:
    # Get the coordinates for the text
    x_coordinate = p.get_x() + p.get_width() / 2  # Center of the bar on the x-axis
    y_coordinate = p.get_height()                  # Top of the bar

    # Format the text string
    # We use an f-string to format the height (the value) as currency.
    # The ':.2f' means format as a float with 2 decimal places.
    text_to_display = f"${y_coordinate:.2f}"

    # Place the text on the plot
    ax.text(x=x_coordinate,
            y=y_coordinate + 0.3, # Add a small offset to lift the text above the bar
            s=text_to_display,    # The string to display
            ha='center',          # Horizontal alignment = center
            fontsize=12)          # Font size


plt.show()

```

```python
import squarify
# squarify.plot()

tips = sns.load_dataset('tips')

# Filter the DataFrame to only include rows where the 'day' is 'Sun'
sunday_data = tips[tips['day'] == 'Sun']

# Get the counts for each value in the 'size' column
party_size_counts = sunday_data['size'].value_counts()


# Get the values for the rectangle sizes
sizes = party_size_counts.values

# Get the index for the labels
labels = party_size_counts.index


label_text = [f'Size {size}\n({count} tables)' for size, count in zip(labels, sizes)]

# --- Full Code for the Treemap ---
import squarify
import matplotlib.pyplot as plt
import seaborn as sns

# Set up the figure and axes
fig, ax = plt.subplots(figsize=(12, 8))

# Define some nice colors
colors = sns.color_palette("viridis", len(sizes))

# Create the treemap
squarify.plot(sizes=sizes,
              label=label_text,
              color=colors,
              alpha=0.8, # Use transparency
              ax=ax)

# Polish the plot
plt.title('Composition of Party Sizes on Sunday (The Most Profitable Day)', fontsize=16, fontweight='bold')
plt.axis('off') # Turn off the x and y axes for a cleaner look
fig, axes = plt.subplots(1, 2, figsize=(20, 8))
ax1 = axes[0,0]
ax2 = axes[1,0]

plt.show()
```
